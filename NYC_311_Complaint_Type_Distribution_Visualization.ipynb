# NYC 311 Complaint Type Distribution Visualization

## Introduction
This Jupyter Notebook provides a comprehensive analysis of complaint types from the NYC 311 data, including visual representations such as bar charts, pie charts, borough-based analysis, and seasonal trends.

## Data Loading


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data = pd.read_csv('nyc_311_data.csv')  # replace with actual dataset path
```



## Data Cleaning


```python
# Preprocessing steps to clean the data
data.dropna(inplace=True)
# Additional cleaning steps...
```



## Visualization Analysis
### Complaint Type Distribution (Bar Chart)


```python
complaint_counts = data['complaint_type'].value_counts()
plt.figure(figsize=(10,6))
sns.barplot(x=complaint_counts.index, y=complaint_counts.values)
plt.title('Distribution of NYC 311 Complaint Types')
plt.xlabel('Complaint Type')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=90)
plt.show()
```



### Complaint Type Distribution (Pie Chart)


```python
plt.figure(figsize=(8,8))
plt.pie(complaint_counts.values, labels=complaint_counts.index, autopct='%1.1f%%')
plt.title('Complaint Type Distribution')
plt.show()
```



### Borough Based Analysis


```python
borough_counts = data['borough'].value_counts()
plt.figure(figsize=(10,6))
sns.barplot(x=borough_counts.index, y=borough_counts.values)
plt.title('Number of Complaints per Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=90)
plt.show()
```



### Seasonal Trends


```python
data['created_date'] = pd.to_datetime(data['created_date'])
data['month'] = data['created_date'].dt.month
seasonal_trends = data.groupby('month')['complaint_type'].count()
plt.figure(figsize=(10,6))
sns.lineplot(x=seasonal_trends.index, y=seasonal_trends.values)
plt.title('Seasonal Trends in NYC 311 Complaints')
plt.xlabel('Month')
plt.ylabel('Number of Complaints')
plt.xticks(range(1, 13))
plt.show()
```



## Conclusion
This notebook provided insights into the complaint types reported to NYC 311, highlighting trends over time and geographical distribution based on boroughs.
